```python
counts=(df["species"].astype(str).str.strip().str.title().value_counts().sort_index())

plt.figure(figsize=(5, 5))
sns.barplot(x=counts.index, y=counts.values)
plt.xlabel('Especies')
plt.ylabel('Frecuencia Absoluta')
plt.title('Tipo de especies')
plt.show()
```

```python
counts=(df["species"].astype(str).str.strip().str.title().value_counts().sort_index())

plt.figure(figsize=(2, 2))
plt.barh(counts.index, counts.values)
plt.xlabel('Especies')
plt.ylabel('Frecuencia Absoluta')
plt.title('Tipo de especies')
plt.show()
```

```python
ax=sns.countplot(data=df, y="species")
ax.set_xlabel('Frecuencia Absoluta')
ax.set_ylabel('Especies')
ax.set_title('Tipo de especies')
plt.show()
```

```python
counts2=(
    df["sex"].astype(str).str.strip().str.title().value_counts()
)
counts2
```

```python
plt.figure(figsize=(5, 6))
plt.pie(counts2.values.tolist(),
        labels=counts2.index.tolist(),
        autopct='%1.1f%%', startangle=90)
plt.title('Sexo')
plt.show()
```

```python
counts3=(
    df["island"].astype(str).str.strip().str.title().value_counts().sort_values()
)
counts3
```

```python
x=range(len(counts3.index))

plt.figure(figsize=(5, 5))
plt.plot(x, counts3.values.tolist(), marker="o")
plt.xticks(x, counts3.index.to_list())
plt.xlabel('Isla')
plt.ylabel('Frecuencia Absoluta')
plt.title('Tipo de isla')
plt.show()
```

```python
counts4=pd.crosstab(
    df['island'].astype(str).str.strip().str.title(),
    df['sex'].astype(str).str.strip().str.title(),
    margins=True
)
counts4
```

```python
counts4=pd.crosstab(
    df['island'].astype(str).str.strip().str.title(),
    df['sex'].astype(str).str.strip().str.title()
)
counts4
```

```python
ax=counts4.plot(kind='bar', stacked=False)
plt.show()
```

```python
ax=counts4.plot(kind='bar', stacked=True)
plt.show()
```

```python
props=pd.crosstab(
    df['species'].astype(str).str.strip().str.title(),
    df['sex'].astype(str).str.strip().str.title(),
    normalize="columns"
)*100
props
```

```python
ax=props.plot(kind='bar', stacked=False)
ax.set_title('Proporción de especies por sexo')
ax.set_ylabel('Proporción (%)')
ax.set_xlabel('Especies')
ax.set_ylim(0, 60)
plt.show()
```

```python
ax=props.plot(kind='line', stacked=False)
ax.set_title('Proporción de especies por sexo')
ax.set_ylabel('Proporción (%)')
ax.set_xlabel('Especies')
ax.set_ylim(0, 60)
plt.show()
```
