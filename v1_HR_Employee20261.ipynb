```python
# Calcular la cantidad de valores atípicos detectados directamente del boxplot
cantidad_outliers2 = len(outliers_detectados2)
cantidad_outliers2
```

```python
# Asegurar que los valores sean conjuntos para comparación rápida
outliers_set1 = set(outliers_detectados)     # Outliers detectados desde la base original
outliers_set2 = set(outliers_detectados2)    # Nuevos outliers detectados desde df_sin_outliers

# Crear la nueva variable que clasifica los datos en 0, 1 o 2
def clasificacion_outlier_completa(valor):
    if valor in outliers_set1:
        return 1  # Primer grupo de outliers
    elif valor in outliers_set2:
        return 2  # Nuevos outliers tras limpieza
    else:
        return 0  # No es outlier

# Aplicar la clasificación
df['Tipo_Outlier_MonthlyIncome'] = df['MonthlyIncome'].apply(clasificacion_outlier_completa)

# Verificar la cantidad de registros por tipo
conteo_final_outliers = df['Tipo_Outlier_MonthlyIncome'].value_counts().sort_index()
conteo_final_outliers
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# 1. Selección de variables
X = num[['JobLevel']]               # Variable independiente (predictora)
y = num['MonthlyIncome']           # Variable dependiente (respuesta)

# 2. Ajustar el modelo lineal simple sobre toda la base
modelo = LinearRegression()
modelo.fit(X, y)

# 3. Predicción sobre los mismos datos
y_pred = modelo.predict(X)

# 4. Evaluación del modelo
intercepto = modelo.intercept_
coeficiente = modelo.coef_[0]
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

# 5. Mostrar resultados
print("=== MODELO LINEAL SIMPLE ===")
print(f"Ecuación: MonthlyIncome = {intercepto:.2f} + {coeficiente:.2f} * JobLevel")
print(f"R² (R-cuadrado): {r2:.4f}")
print(f"MSE: {mse:.2f}")

# 6. Gráfico de dispersión + línea de regresión
plt.figure(figsize=(8, 6))
plt.scatter(X, y, color='blue', label='Datos reales')
plt.plot(X, y_pred, color='red', linewidth=2, label='Línea de regresión')
plt.title('Modelo Lineal Simple: MonthlyIncome vs JobLevel')
plt.xlabel('JobLevel')
plt.ylabel('MonthlyIncome')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

# **PCA: Analisis de componentes principales**

```python
from matplotlib.patches import Ellipse
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
```

```python
# 1. Seleccionar solo variables numéricas y estandarizarlas
num = df.select_dtypes(exclude='O')
num_scaled = StandardScaler().fit_transform(num)

# 2. Aplicar PCA (conservar todos los componentes para análisis)
pca = PCA()
pca_fit = pca.fit(num_scaled)
varianza_explicada = pca_fit.explained_variance_ratio_

# 3. Crear un DataFrame con la varianza explicada por cada componente
pca_resultados = pd.DataFrame({
    'Componente Principal': [f'PC{i+1}' for i in range(len(varianza_explicada))],
    'Varianza Explicada (%)': varianza_explicada * 100,
    'Varianza Acumulada (%)': varianza_explicada.cumsum() * 100
})
print(pca_resultados)
```

```python
# Ajustar el código original para mostrar la gráfica del Scree Plot directamente
plt.figure(figsize=(10, 6))
plt.plot(range(1, len(varianza_explicada) + 1), varianza_explicada * 100, marker='o', linestyle='-')
plt.title('Scree Plot: Varianza Explicada por Componente Principal')
plt.xlabel('Número de Componentes Principales')
plt.ylabel('Varianza Explicada (%)')
plt.xticks(range(1, len(varianza_explicada) + 1))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Función para visualizar el círculo de correlaciones para cualquier par de componentes seleccionados
def graficar_circulo_correlaciones(num_scaled, num_columns, componente_x=1, componente_y=2):
    from matplotlib.patches import Circle

    # Ajustar PCA con al menos hasta el componente máximo solicitado
    pca = PCA(n_components=max(componente_x, componente_y))
    pca.fit(num_scaled)
    cargas = pca.components_.T  # columnas = PC1, PC2, ..., PCn

    # Seleccionar las dos componentes deseadas (ajustar a índice 0)
    x_idx = componente_x - 1
    y_idx = componente_y - 1

    # Crear gráfico
    fig, ax = plt.subplots(figsize=(8, 8))

    # Dibujar círculo de referencia
    circle = Circle((0, 0), 1, facecolor='none', edgecolor='grey', linestyle='--')
    ax.add_patch(circle)

    # Ejes
    ax.axhline(0, color='black', linestyle='--')
    ax.axvline(0, color='black', linestyle='--')

    # Flechas y etiquetas
    for i, var in enumerate(num_columns):
        plt.arrow(0, 0, cargas[i, x_idx], cargas[i, y_idx],
                  head_width=0.03, head_length=0.05, fc='red', ec='red')
        plt.text(cargas[i, x_idx]*1.15, cargas[i, y_idx]*1.15, var, fontsize=9, ha='center', va='center')

    # Etiquetas ejes y título
    ax.set_xlim(-1.1, 1.1)
    ax.set_ylim(-1.1, 1.1)
    ax.set_xlabel(f"Dim{componente_x} ({pca.explained_variance_ratio_[x_idx]*100:.1f}%)")
    ax.set_ylabel(f"Dim{componente_y} ({pca.explained_variance_ratio_[y_idx]*100:.1f}%)")
    ax.set_title(f"Círculo de correlaciones - PCA (Dim{componente_x} vs Dim{componente_y})")
    plt.grid(False)
    plt.gca().set_aspect('equal', adjustable='box')
    plt.tight_layout()
    plt.show()

# Ejecutar la función con valores por defecto PC1 vs PC2
graficar_circulo_correlaciones(num_scaled, num.columns, componente_x=1, componente_y=2)
```

```python
# Definir función para elipses
def dibujar_elipse(ax, data, n_std=2.0, facecolor='none', **kwargs):
    if data.shape[0] < 2:
        return
    cov = np.cov(data.T)
    mean = np.mean(data, axis=0)
    vals, vecs = np.linalg.eigh(cov)
    order = vals.argsort()[::-1]
    vals = vals[order]
    vecs = vecs[:, order]
    theta = np.degrees(np.arctan2(*vecs[:, 0][::-1]))
    width, height = 2 * n_std * np.sqrt(vals)
    el = Ellipse(xy=mean, width=width, height=height, angle=theta,
                 facecolor=facecolor, **kwargs)
    ax.add_patch(el)

# PCA con género
pca_genero = PCA(n_components=2)
componentes_genero = pca_genero.fit_transform(num_scaled)
df_pca_genero = pd.DataFrame(componentes_genero, columns=['PC1', 'PC2'])
df_pca_genero['Grupo'] = df['Gender']
Genero = df['Gender'].unique()
colores = dict(zip(Genero, sns.color_palette("Set2", len(Genero))))

plt.figure(figsize=(10, 7))
ax = plt.gca()
for grupo, datos in df_pca_genero.groupby('Grupo'):
    color = colores[grupo]
    plt.scatter(datos['PC1'], datos['PC2'], label=grupo, color=color, alpha=0.6, s=60)
    dibujar_elipse(ax, datos[['PC1', 'PC2']].values, facecolor=color, alpha=0.1, edgecolor=color)
plt.axhline(0, color='black', linestyle='--')
plt.axvline(0, color='black', linestyle='--')
plt.xlabel(f'Dim1 ({pca_genero.explained_variance_ratio_[0]*100:.1f}%)')
plt.ylabel(f'Dim2 ({pca_genero.explained_variance_ratio_[1]*100:.1f}%)')
plt.title('Individuos - PCA con Elipses por Grupo (Gender)')
plt.legend(title='Gender')
plt.tight_layout()
plt.show()
```

```python
# Función para mostrar dendrograma y asignar grupos a las variables
def agrupar_variables_por_dendrograma(n_grupos=3):
    # Calcular y graficar dendrograma
    Z_var = linkage(num_scaled.T, method='ward')

    plt.figure(figsize=(12, 6))
    dendrogram(Z_var, labels=num.columns, leaf_rotation=90., leaf_font_size=10.)
    plt.title(f"Dendrograma Jerárquico de Variables (k={n_grupos})")
    plt.xlabel("Variables")
    plt.ylabel("Distancia euclidiana")
    plt.tight_layout()
    plt.show()

    # Cortar el árbol para obtener grupos
    grupos = fcluster(Z_var, t=n_grupos, criterion='maxclust')
    agrupamiento = pd.DataFrame({'Variable': num.columns, 'Grupo': grupos})
    return agrupamiento.sort_values(by='Grupo')

# Ejecutar con número de grupos modificable (ejemplo: 3)
agrupar_variables_por_dendrograma(n_grupos=3)
```
