
# Análisis de Regresión Lineal Simple y Regresión Múltiple

Antes de aplicar PCA y clustering, se incorpora un análisis de regresión para estudiar relaciones lineales entre el índice de felicidad y sus principales factores explicativos.

## Variable respuesta

- `Ladder score`: puntuación de felicidad o bienestar subjetivo.

## Regresión lineal simple

Se evalúa la relación entre:

- `Ladder score`
- `Log GDP per capita`

## Regresión lineal múltiple

Se evalúa la relación conjunta entre `Ladder score` y los siguientes predictores:

- `Log GDP per capita`
- `Social support`
- `Healthy life expectancy`
- `Freedom to make life choices`
- `Generosity`
- `Perceptions of corruption`


```python

# ============================================================
# Preparación de variables para regresión
# ============================================================

import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.stats.outliers_influence import variance_inflation_factor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split

# Columnas reales del World Happiness Report 2024
regression_columns = [
    "Ladder score",
    "Log GDP per capita",
    "Social support",
    "Healthy life expectancy",
    "Freedom to make life choices",
    "Generosity",
    "Perceptions of corruption"
]

# Se verifica que las columnas existan en la base
missing_cols = [col for col in regression_columns if col not in df.columns]
if missing_cols:
    raise ValueError(f"Estas columnas no están en la base: {missing_cols}")

df_reg = df[regression_columns + ["Country name", "Regional indicator"]].dropna().copy()

# Renombrar columnas para evitar problemas con espacios en fórmulas
df_reg = df_reg.rename(columns={
    "Ladder score": "Ladder_score",
    "Log GDP per capita": "Log_GDP_per_capita",
    "Social support": "Social_support",
    "Healthy life expectancy": "Healthy_life_expectancy",
    "Freedom to make life choices": "Freedom_to_make_life_choices",
    "Perceptions of corruption": "Perceptions_of_corruption",
    "Country name": "Country_name",
    "Regional indicator": "Regional_indicator"
})

print("Dimensiones de la base para regresión:", df_reg.shape)
display(df_reg.head())
```


## 1. Regresión lineal simple

Modelo estimado:

$$
Ladder\_score = \beta_0 + \beta_1 Log\_GDP\_per\_capita + \epsilon
$$

Este modelo permite estudiar si los países con mayor PIB per cápita logarítmico tienden a presentar mayor puntuación de felicidad.


```python

# ============================================================
# Regresión lineal simple
# ============================================================

simple_model = smf.ols(
    "Ladder_score ~ Log_GDP_per_capita",
    data=df_reg
).fit()

print(simple_model.summary())
```

```python

# ============================================================
# Gráfico de regresión lineal simple
# ============================================================

plt.figure(figsize=(8, 5))
sns.regplot(
    data=df_reg,
    x="Log_GDP_per_capita",
    y="Ladder_score",
    scatter_kws={"alpha": 0.75}
)
plt.title("Regresión lineal simple: Ladder score vs Log GDP per capita")
plt.xlabel("Log GDP per capita")
plt.ylabel("Ladder score")
plt.show()
```

```python

# ============================================================
# Métricas del modelo simple
# ============================================================

y_simple = df_reg["Ladder_score"]
y_simple_pred = simple_model.fittedvalues

simple_metrics = pd.DataFrame({
    "Modelo": ["Regresión lineal simple"],
    "R2": [r2_score(y_simple, y_simple_pred)],
    "RMSE": [np.sqrt(mean_squared_error(y_simple, y_simple_pred))],
    "MAE": [mean_absolute_error(y_simple, y_simple_pred)]
})

display(simple_metrics.round(4))
```


## 2. Regresión lineal múltiple

Modelo estimado:

$$
Ladder\_score =
\beta_0 +
\beta_1 Log\_GDP\_per\_capita +
\beta_2 Social\_support +
\beta_3 Healthy\_life\_expectancy +
\beta_4 Freedom\_to\_make\_life\_choices +
\beta_5 Generosity +
\beta_6 Perceptions\_of\_corruption +
\epsilon
$$

Este modelo permite evaluar qué factores mantienen asociación con la felicidad cuando se controlan simultáneamente los demás predictores.


```python

# ============================================================
# Regresión lineal múltiple
# ============================================================

multiple_model = smf.ols(
    "Ladder_score ~ Log_GDP_per_capita + Social_support + Healthy_life_expectancy + Freedom_to_make_life_choices + Generosity + Perceptions_of_corruption",
    data=df_reg
).fit()

print(multiple_model.summary())
```

```python

# ============================================================
# Métricas del modelo múltiple
# ============================================================

y_multiple = df_reg["Ladder_score"]
y_multiple_pred = multiple_model.fittedvalues

multiple_metrics = pd.DataFrame({
    "Modelo": ["Regresión lineal múltiple"],
    "R2": [r2_score(y_multiple, y_multiple_pred)],
    "R2_ajustado": [multiple_model.rsquared_adj],
    "RMSE": [np.sqrt(mean_squared_error(y_multiple, y_multiple_pred))],
    "MAE": [mean_absolute_error(y_multiple, y_multiple_pred)]
})

display(multiple_metrics.round(4))
```

```python

# ============================================================
# Comparación de modelos
# ============================================================

comparison = pd.concat([
    simple_metrics.assign(R2_ajustado=simple_model.rsquared_adj),
    multiple_metrics
], ignore_index=True)

comparison = comparison[["Modelo", "R2", "R2_ajustado", "RMSE", "MAE"]]
display(comparison.round(4))
```

```python

# ============================================================
# Diagnóstico de residuos del modelo múltiple
# ============================================================

residuals = multiple_model.resid
fitted = multiple_model.fittedvalues

plt.figure(figsize=(8, 5))
sns.scatterplot(x=fitted, y=residuals)
plt.axhline(0, linestyle="--")
plt.title("Modelo múltiple: residuos vs valores ajustados")
plt.xlabel("Valores ajustados")
plt.ylabel("Residuos")
plt.show()

sm.qqplot(residuals, line="45", fit=True)
plt.title("Modelo múltiple: QQPlot de residuos")
plt.show()

plt.figure(figsize=(8, 5))
sns.histplot(residuals, kde=True)
plt.title("Modelo múltiple: distribución de residuos")
plt.xlabel("Residuos")
plt.show()
```

```python

# ============================================================
# Valores reales vs valores predichos
# ============================================================

plt.figure(figsize=(7, 6))
sns.scatterplot(x=y_multiple, y=y_multiple_pred)
plt.plot(
    [y_multiple.min(), y_multiple.max()],
    [y_multiple.min(), y_multiple.max()],
    linestyle="--"
)
plt.title("Modelo múltiple: valores reales vs predichos")
plt.xlabel("Ladder score real")
plt.ylabel("Ladder score predicho")
plt.show()
```

```python

# ============================================================
# Coeficientes del modelo múltiple
# ============================================================

coef_table = pd.DataFrame({
    "Variable": multiple_model.params.index,
    "Coeficiente": multiple_model.params.values,
    "p_value": multiple_model.pvalues.values,
    "IC_2.5%": multiple_model.conf_int()[0].values,
    "IC_97.5%": multiple_model.conf_int()[1].values
})

display(coef_table.round(4))

coef_plot = coef_table[coef_table["Variable"] != "Intercept"].copy()

plt.figure(figsize=(8, 5))
sns.barplot(data=coef_plot, x="Coeficiente", y="Variable")
plt.axvline(0, linestyle="--")
plt.title("Coeficientes estimados del modelo múltiple")
plt.xlabel("Coeficiente")
plt.ylabel("Variable")
plt.show()
```

En regresión múltiple, si dos o más variables explicativas están altamente correlacionadas:

* los coeficientes pueden volverse inestables,
* los errores estándar aumentan,
* las interpretaciones se vuelven poco confiables,
el modelo pierde estabilidad.

Por eso se calcula el VIF.

```python

# ============================================================
# Multicolinealidad: VIF
# ============================================================

X_vif = df_reg[
    [
        "Log_GDP_per_capita",
        "Social_support",
        "Healthy_life_expectancy",
        "Freedom_to_make_life_choices",
        "Generosity",
        "Perceptions_of_corruption"
    ]
].copy()

X_vif_const = sm.add_constant(X_vif)

vif_table = pd.DataFrame({
    "Variable": X_vif_const.columns,
    "VIF": [variance_inflation_factor(X_vif_const.values, i) for i in range(X_vif_const.shape[1])]
})

display(vif_table.round(4))
```

| VIF   | Interpretación           |
| ----- | ------------------------ |
| 1     | No hay multicolinealidad |
| 1 – 3 | Multicolinealidad baja   |
| 3 – 5 | Moderada                 |
| > 5   | Alta                     |
| > 10  | Muy problemática         |


| Variable                     | VIF    | Interpretación             |
| ---------------------------- | ------ | -------------------------- |
| Log_GDP_per_capita           | 4.1573 | Multicolinealidad moderada |
| Social_support               | 2.6256 | Baja                       |
| Healthy_life_expectancy      | 3.5090 | Moderada                   |
| Freedom_to_make_life_choices | 1.4547 | Muy baja                   |
| Generosity                   | 1.1463 | Prácticamente inexistente  |
| Perceptions_of_corruption    | 1.4099 | Muy baja                   |


El análisis del Factor de Inflación de la Varianza (VIF) evidencia que el modelo de regresión múltiple no presenta problemas severos de multicolinealidad. Ninguna variable supera el umbral crítico de 5, lo que sugiere estabilidad en la estimación de los coeficientes. Las variables Log_GDP_per_capita y Healthy_life_expectancy presentan niveles moderados de correlación con otras variables del modelo, posiblemente debido a la relación estructural entre desarrollo económico y condiciones de salud en los países analizados. Sin embargo, los valores observados no comprometen la validez estadística del modelo.

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Ensure features are defined (from previous context)
features = ["Ladder score", "Log GDP per capita", "Social support", "Healthy life expectancy"]

# Prepare data for PCA (as done in previous cells)
df_pca = df_1[features].dropna()

# Scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_pca)

# Apply PCA
pca = PCA()
pca.fit(X_scaled)

# Get eigenvalues (explained variances)
eigenvalues = pca.explained_variance_

# Crear un Scree Plot (gráfico de eigenvalues) profesional
plt.figure(figsize=(10, 6))
plt.plot(range(1, len(eigenvalues)+1), eigenvalues, marker='o', linestyle='-', color='royalblue')
plt.title('Scree Plot - Análisis de Componentes Principales', fontsize=18, fontweight='bold')
plt.xlabel('Número de Componente Principal', fontsize=14)
plt.ylabel('Valor Propio (Eigenvalue)', fontsize=14)
plt.xticks(range(1, len(eigenvalues)+1))
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Calcular porcentaje de varianza explicada
explained_variance_ratio = pca.explained_variance_ratio_ * 100

# Gráfico combinado (barras + línea) con etiquetas de porcentaje
plt.figure(figsize=(10, 6))
bars = plt.bar(range(1, len(explained_variance_ratio)+1), explained_variance_ratio, color='steelblue', alpha=0.9)
plt.plot(range(1, len(explained_variance_ratio)+1), explained_variance_ratio, color='black', marker='o')

# Etiquetas de porcentaje encima de cada barra
for i, var in enumerate(explained_variance_ratio):
    plt.text(i+1, var + 1, f"{var:.1f}%", ha='center', va='bottom', fontsize=10)

# Títulos y ejes
plt.title("Scree Plot - Porcentaje de Varianza Explicada", fontsize=16, fontweight='bold')
plt.xlabel("Componentes Principales", fontsize=12)
plt.ylabel("Porcentaje de Varianza Explicada", fontsize=12)
plt.xticks(range(1, len(explained_variance_ratio)+1))
plt.yticks(range(0, 90, 10))
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
from sklearn.cluster import KMeans
import plotly.graph_objects as go

#Calcular inercia para diferentes k
inertias = []
K_range = range(1, 10)

for k in K_range:
    kmeans = KMeans(n_clusters=k, n_init=10, random_state=42)
    kmeans.fit(X_pca)
    inertias.append(kmeans.inertia_)

# Graficar con Plotly
fig = go.Figure()
fig.add_trace(go.Scatter(
    x=list(K_range),
    y=inertias,
    mode='lines+markers',
    line=dict(color='royalblue'),
    marker=dict(size=8)
))

fig.update_layout(
    title="Método del Codo - Selección del Número Óptimo de Clusters",
    xaxis_title="Número de Clusters (k)",
    yaxis_title="Inercia (Suma de distancias internas)",
    template="plotly_white"
)

fig.show()
```

```python
# Crear un dendrograma con nombres de países rotados 90 grados y selección de clusters (corte)

from scipy.cluster.hierarchy import dendrogram, linkage, fcluster # Import linkage here

# Generar etiquetas de países desde el DataFrame original alineado con PCA
labels = df.loc[df_pca.index, "Country name"].values

# Crear linkage nuevamente para asegurar consistencia
linked = linkage(X_scaled, method='ward') # Use X_scaled instead of pca

# Generar los clusters
cluster_labels = fcluster(linked, n_clusters, criterion='maxclust')

# Crear un dendrograma con colores que representen visualmente los 4 clusters seleccionados

plt.figure(figsize=(17, 10))
dendrogram(
    linked,
    labels=labels,
    orientation='top',
    leaf_rotation=90,
    leaf_font_size=10,
    color_threshold=linked[-(n_clusters-1), 2],  # Corta en la altura que genera n_clusters
    above_threshold_color='gray'  # Color para ramas superiores al corte
)

plt.title('Endograma con Colores para 4 Clusters Jerárquicos', fontsize=18, fontweight='bold')
plt.xlabel('País')
plt.ylabel('Distancia Euclidiana')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Volver a crear el boxplot de clusters con índice de felicidad y resaltar los países atípicos

plt.figure(figsize=(12, 6))
sns.boxplot(data=df_box, x='Cluster_HC', y='Ladder score', palette='Set2')

# Calcular los outliers para cada cluster
def get_outliers(group):
    Q1 = group.quantile(0.25)
    Q3 = group.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return group[(group < lower_bound) | (group > upper_bound)]

outliers = df_box.groupby('Cluster_HC')['Ladder score'].apply(get_outliers).droplevel(0)
outliers_df = df_box[df_box.index.isin(outliers.index)]

# Resaltar los países atípicos
for i in outliers_df.index:
    x = df_box.loc[i, 'Cluster_HC'] - 1  # eje x: cluster (indexado desde 0)
    y = df_box.loc[i, 'Ladder score']
    label = df_box.loc[i, 'Country name']
    plt.scatter(x, y, color='red', s=50, edgecolor='black', zorder=5)
    plt.text(x + 0.05, y, label, fontsize=9, color='red', ha='left')

plt.title('Distribución del Índice de Felicidad por Cluster Jerárquico (con Outliers)', fontsize=16, fontweight='bold')
plt.xlabel('Cluster Jerárquico')
plt.ylabel('Índice de Felicidad (Ladder Score)')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
