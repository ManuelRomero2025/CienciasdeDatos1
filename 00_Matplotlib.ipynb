# **Matplotlib**

Matplotlib es una biblioteca de Python para visualización de datos que permite crear gráficos estáticos, animados e interactivos. Es una de las herramientas más utilizadas en ciencia de datos, estadística y machine learning para representar información de manera gráfica.

Características principales de Matplotlib

* Flexibilidad: permite crear desde gráficos básicos (líneas, barras, histogramas, dispersión) hasta visualizaciones más complejas.

* Control detallado: se pueden personalizar colores, estilos de líneas, fuentes, leyendas, títulos, etiquetas de ejes, etc.

* Compatibilidad: se integra fácilmente con NumPy, pandas y otras bibliotecas de ciencia de datos.

* Formatos de salida: exporta gráficos en múltiples formatos (PNG, PDF, SVG, etc.).

* Módulo pyplot: el submódulo matplotlib.pyplot se usa para crear gráficos de forma sencilla, similar a MATLAB.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Datos de ejemplo
x = [1, 2, 3, 4, 5]
y = [2, 4, 7, 1, 6]

# Crear gráfico de línea
plt.plot(x, y, marker='o', color='blue', linestyle='--', label="Serie de datos")

# Personalizar
plt.title("Ejemplo de gráfico con Matplotlib")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.legend()
plt.grid(True)

# Mostrar gráfico
plt.show()
```

```python
# ---------------------------
# 2. Barras verticales (plt.bar)
# ---------------------------
categorias = ["A", "B", "C", "D"]
valores = [5, 7, 3, 8]

plt.bar(categorias, valores, color="orange")
plt.title("Gráfico de barras verticales")
plt.xlabel("Categorías")
plt.ylabel("Valores")
plt.show()
```

```python
# ---------------------------
# 3. Barras horizontales (plt.barh)
# ---------------------------
plt.barh(categorias, valores, color="green")
plt.title("Gráfico de barras horizontales")
plt.xlabel("Valores")
plt.ylabel("Categorías")
plt.show()
```

```python
# ---------------------------
# 4. Histograma (plt.hist)
# ---------------------------
datos = [7,8,5,6,9,5,6,7,8,9,10,6,7,8,5,9,10,11,7,8]

plt.hist(datos, bins=5, color="purple", edgecolor="black")
plt.title("Histograma")
plt.xlabel("Intervalos")
plt.ylabel("Frecuencia")
plt.show()
```

```python
# ---------------------------
# 5. Diagrama de dispersión (plt.scatter)
# ---------------------------
x = [5, 7, 8, 7, 6, 9, 5, 6, 7, 8]
y = [99, 86, 87, 88, 100, 86, 103, 87, 94, 78]

plt.scatter(x, y, color="red")
plt.title("Diagrama de dispersión")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
# ---------------------------
# 6. Gráfico circular (plt.pie)
# ---------------------------
valores = [40, 25, 20, 15]
etiquetas = ["A", "B", "C", "D"]
colores = ["gold", "lightblue", "lightcoral", "lightgreen"]

plt.pie(valores, labels=etiquetas, colors=colores, autopct="%1.1f%%", startangle=90)
plt.title("Gráfico circular")
plt.show()
```

```python
# Conteo por especie (sin transformar más allá de título-caso)
counts = (
    df["species"]
    .astype(str).str.strip().str.title()
    .value_counts()
    .sort_index()
)

plt.figure()
plt.bar(counts.index.tolist(), counts.values.tolist())
plt.title("Species (conteo)")
plt.xlabel("Species")
plt.ylabel("Conteo")
plt.tight_layout()
plt.show()
```

```python
counts = (
    df["island"]
    .astype(str).str.strip().str.title()
    .value_counts()
    .sort_index()
)

plt.figure()
plt.barh(counts.index.tolist(), counts.values.tolist())
plt.title("Island (conteo)")
plt.xlabel("Conteo")
plt.ylabel("Island")
plt.tight_layout()
plt.show()
```

```python
counts = (
    df["sex"]
    .astype(str).str.strip().str.title()
    .value_counts()
)

plt.figure()
plt.pie(counts.values.tolist(), labels=counts.index.tolist(),
        autopct="%1.1f%%", startangle=90)
plt.title("Sex (porcentaje)")
plt.tight_layout()
plt.show()
```

```python
counts = (
    df["species"]
    .astype(str).str.strip().str.title()
    .value_counts()
    .sort_index()
)

x = range(len(counts))

plt.figure()
plt.plot(x, counts.values.tolist(), marker="o")
plt.xticks(x, counts.index.tolist())
plt.title("Species (conteo) — Línea")
plt.xlabel("Species")
plt.ylabel("Conteo")
plt.tight_layout()
plt.show()
```

```python
# Tabla island x species
counts = pd.crosstab(
    df["island"].astype("string").str.strip().str.title(),
    df["species"].astype("string").str.strip().str.title()
)

# Barras agrupadas + etiquetas
ax = counts.plot(kind="bar")
ax.set_title("Species vs Island")
ax.set_xlabel("Island")
ax.set_ylabel("Conteo")
ax.legend(title="Species")

for c in ax.containers:
    ax.bar_label(c, fmt="%d", padding=2)

plt.tight_layout()
plt.show()
```

```python
# Tabla island x species (normalizada por fila -> proporción por isla)
props = pd.crosstab(
    df["island"].astype("string").str.strip().str.title(),
    df["species"].astype("string").str.strip().str.title(),
    normalize="index"
) * 100  # a porcentaje

# Barras agrupadas + etiquetas (porcentaje)
ax = props.plot(kind="bar")
ax.set_title("Species vs Island — Proporción (%)")
ax.set_xlabel("Island")
ax.set_ylabel("Proporción (%)")
ax.set_ylim(0, 100)

for container in ax.containers:
    ax.bar_label(container, labels=[f"{v:.1f}%" for v in container.datavalues], padding=2)

plt.tight_layout()
```

```python
plt.figure()
plt.hist(df["body_mass_g"], bins=10, rwidth=0.9)
plt.title("Histograma de body_mass_g")
plt.xlabel("Masa corporal (g)")
plt.ylabel("Frecuencia")
plt.tight_layout()
plt.show()
```

```python
# Boxplot de una sola variable
plt.figure()
plt.boxplot(df["body_mass_g"].dropna(), labels=["body_mass_g"])
plt.title("Boxplot de body_mass_g")
plt.ylabel("Masa corporal (g)")
plt.tight_layout()
```

```python
# Boxplot de una sola variable con color SOLO en la caja
plt.figure()
plt.boxplot(
    df["body_mass_g"].dropna(),
    labels=["body_mass_g"],
    patch_artist=True,                # habilita color de relleno
    boxprops=dict(facecolor="#cfe8ff")# color de fondo SOLO de la caja
)
plt.title("Boxplot de body_mass_g")
plt.ylabel("Masa corporal (g)")
plt.tight_layout()
plt.show()
```

```python

df["island"] = df["island"].astype(str).str.strip().str.title()
df = df.dropna(subset=["island", "body_mass_g"])

# Arreglos por isla
islands = sorted(df["island"].unique())
data = [df.loc[df["island"] == isl, "body_mass_g"].values for isl in islands]

# Boxplot
plt.figure()
plt.boxplot(data, labels=islands)
plt.title("Boxplot de body_mass_g por island")
plt.xlabel("Island")
plt.ylabel("Masa corporal (g)")
plt.tight_layout()
```
